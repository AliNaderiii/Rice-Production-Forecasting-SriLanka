We run a dual-panel analysis for each continuous variable. Visual histograms with overlaid KDE help inspect skewed patterns, 
while normal Q-Q plots confirm how close features are to Gaussian assumptions.
</div>


```python
# tag: ext_stats
ext_stats = []
for col in CONTINUOUS + [TARGET]:
    x = df[col].dropna()
    ext_stats.append({
        "Feature"  : col,
        "Mean"     : round(x.mean(), 2),
        "Median"   : round(x.median(), 2),
        "Std"      : round(x.std(), 2),
        "Skewness" : round(float(x.skew()), 3),
        "Kurtosis" : round(float(kurtosis(x.values)), 3),
        "CV (%)"   : round(x.std() / x.mean() * 100, 1),
    })
ext_df = pd.DataFrame(ext_stats).set_index("Feature")
ext_df
```

```python
# tag: distributions
cols_to_plot = CONTINUOUS + [TARGET]
fig = plt.figure(figsize=(20, 24), facecolor=DARK_BG)
fig.suptitle("Feature Distribution and Normality (Q-Q) Audit", fontsize=14, color="white", y=1.00, fontweight="bold")

for i, col in enumerate(cols_to_plot):
    x = df[col].dropna().values
    mu = np.mean(x); med = np.median(x); sd = np.std(x)
    sk = float(stats.skew(x)); ku = float(kurtosis(x))

    ax_h = fig.add_subplot(7, 2, 2*i + 1)
    ax_q = fig.add_subplot(7, 2, 2*i + 2)

    # Histogram + KDE
    ax_h.hist(x, bins=20, color=FIELD_GRN, alpha=0.6, edgecolor=DARK_BG, linewidth=0.5, density=True)
    xr = np.linspace(x.min(), x.max(), 300)
    kde = stats.gaussian_kde(x)
    ax_h.plot(xr, kde(xr), color=GOLD_GLOW, lw=2, label="KDE")
    ax_h.axvline(mu, color=WARN_RED, lw=1.5, ls="--", label=f"Mean={mu:.1f}")
    ax_h.axvline(med, color="white", lw=1.2, ls=":", label=f"Median={med:.1f}")
    ax_h.axvspan(mu-sd, mu+sd, alpha=0.08, color=GOLD_GLOW)

    stats_txt = "\n".join([
        f"Std  = {sd:.2f}",
        f"Skew = {sk:.2f}",
        f"Kurt = {ku:.2f}",
        f"CV   = {sd/mu*100:.1f}%"
    ])
    ax_h.text(0.97, 0.96, stats_txt, transform=ax_h.transAxes, fontsize=8.5,
              verticalalignment="top", horizontalalignment="right",
              bbox=dict(boxstyle="round,pad=0.5", facecolor=DARK_BG, edgecolor=GOLD, linewidth=1),
              color="#E2ECD8", family="monospace")
    ax_h.legend(fontsize=8.5, framealpha=0.3)
    ax_h.set_title(f"{col.upper()}  —  Histogram + KDE", fontsize=10.5)
    ax_h.set_xlabel(col, fontsize=8.5); ax_h.set_ylabel("Density", fontsize=8.5)
    ax_h.set_facecolor(CARD_BG)

    # Q-Q Plot
    (osm, osr), (slope, intercept, r_val) = probplot(x, dist="norm")
    ax_q.scatter(osm, osr, s=15, color=FIELD_GRN, alpha=0.6, edgecolors="none")
    qq_line = slope * np.array([osm[0], osm[-1]]) + intercept
    ax_q.plot([osm[0], osm[-1]], qq_line, color=GOLD_GLOW, lw=1.8, label=f"R={r_val:.3f}")
    verdict = "Approx Normal" if abs(sk) < 0.5 else ("Right Skewed" if sk > 0 else "Left Skewed")
    ax_q.text(0.03, 0.95, verdict, transform=ax_q.transAxes, fontsize=9,
              color=GOLD_GLOW if abs(sk) < 0.5 else WARN_RED, fontweight="bold", verticalalignment="top")
    ax_q.legend(fontsize=8.5, framealpha=0.3)
    ax_q.set_title(f"{col.upper()}  —  Q-Q Plot", fontsize=10.5)
    ax_q.set_xlabel("Theoretical Quantiles", fontsize=8.5)
    ax_q.set_ylabel("Sample Quantiles", fontsize=8.5)
    ax_q.set_facecolor(CARD_BG)

plt.tight_layout(h_pad=2.2)
plt.show()
```

Because relationships in agricultural and meteorological data are often non-linear, we utilize 
<strong style="color:#FFF">Spearman Rank Correlation</strong> instead of Pearson. 
This quantifies monotonic relationships safely without assuming linearity.

```python
# tag: correlation
# Calculate spearman correlation matrix
numeric_cols = CONTINUOUS + [TARGET]
corr_mat, p_values = spearmanr(df[numeric_cols])
corr_df = pd.DataFrame(corr_mat, index=numeric_cols, columns=numeric_cols)

# Print targeted correlation with production
target_corr = corr_df[TARGET].drop(TARGET).sort_values()
print("Spearman Rank Correlation with Production:")
print(target_corr.to_string())

fig, axes = plt.subplots(1, 2, figsize=(18, 7.5), facecolor=DARK_BG)
mask = np.triu(np.ones_like(corr_df, dtype=bool))
custom_cmap = LinearSegmentedColormap.from_list("wheat_cmap", [WARN_RED, CARD_BG, GOLD], N=256)

# Heatmap
sns.heatmap(corr_df, mask=mask, cmap=custom_cmap, annot=True, fmt=".2f",
            center=0, vmin=-1, vmax=1, ax=axes[0],
            annot_kws={"size":8.5, "color":"white"},
            cbar_kws={"shrink": 0.8, "label": "Spearman Rank r"})
axes[0].set_title("Spearman Rank Correlation Matrix", fontsize=12, pad=12)
axes[0].set_facecolor(CARD_BG)

# Horizontal bar correlation with target
colors = [WARN_RED if val < 0 else GOLD for val in target_corr.values]
bars = axes[1].barh(target_corr.index, target_corr.values, color=colors, height=0.55, edgecolor=DARK_BG)
axes[1].axvline(0, color="#2E3B27", lw=1)
for bar, val in zip(bars, target_corr.values):
    axes[1].text(val + (0.015 if val>=0 else -0.015), bar.get_y()+bar.get_height()/2,
                 f"{val:+.3f}", va="center", ha="left" if val>=0 else "right",
                 color="white", fontweight="bold", fontsize=9)
axes[1].set_xlim(-0.65, 1.1)
axes[1].set_title("Feature Correlations with target Production (Monotonic)", fontsize=12, pad=12)
axes[1].set_xlabel("Spearman r")
axes[1].set_facecolor(CARD_BG)

plt.tight_layout()
plt.show()
```

<div style="background:rgba(255,176,0,0.05);border:1px solid rgba(255,176,0,0.2);border-radius:10px;padding:16px 22px;font-family:'Segoe UI',system-ui,sans-serif;font-size:14px;color:#cbdcb8;line-height:1.8">
We evaluate the stationarity of our target series using the **Augmented Dickey-Fuller (ADF)** test. 
This determines if the series requires differencing (d=1) to prevent spurious correlations in the ARIMA model.
</div>


```python
# tag: mannwhitney
log_prod = np.log(df[TARGET])
res_raw  = adfuller(log_prod.values)
res_diff = adfuller(log_prod.diff().dropna().values)

adf_data = [
    {"Series": "log(Production)", "ADF Stat": round(res_raw[0],4), "p-value": f"{res_raw[1]:.2e}", "Stationary": "No"},
    {"Series": "1st Diff log(Production)", "ADF Stat": round(res_diff[0],4), "p-value": f"{res_diff[1]:.2e}", "Stationary": "Yes"}
]
print(pd.DataFrame(adf_data).to_string(index=False))

# Plotting original vs differenced series
fig, axes = plt.subplots(2, 1, figsize=(14, 8), facecolor=DARK_BG)
axes[0].plot(df["Year_New"] + (df["Season"]=="Maha")*0.5, log_prod, color=GOLD, label="log(Production)")
axes[0].set_title("Original Log Production Series (1950-2024)", fontsize=11.5)
axes[0].legend(); axes[0].set_facecolor(CARD_BG)

axes[1].plot(df["Year_New"].iloc[1:] + (df["Season"].iloc[1:]=="Maha")*0.5, log_prod.diff().dropna(), color=FIELD_GRN, label="1st Diff log(Production)")
axes[1].set_title("First Differenced Log Production Series (Stationary)", fontsize=11.5)
axes[1].legend(); axes[1].set_facecolor(CARD_BG)

plt.tight_layout()
plt.show()
```

```python
# tag: outliers
# Outlier Check using IQR method
Q1 = df[CONTINUOUS].quantile(0.25)
Q3 = df[CONTINUOUS].quantile(0.75)
IQR = Q3 - Q1
lo = Q1 - 1.5 * IQR; hi = Q3 + 1.5 * IQR
outlier_mask = (df[CONTINUOUS] < lo) | (df[CONTINUOUS] > hi)

out_info = []
for col in CONTINUOUS:
    n_out = outlier_mask[col].sum()
    out_info.append({
        "Feature": col, "Q1": round(Q1[col],2), "Q3": round(Q3[col],2),
        "IQR": round(IQR[col],2), "Outliers": n_out, "Pct%": f"{n_out/len(df)*100:.1f}%"
    })
out_df = pd.DataFrame(out_info).set_index("Feature")
print(out_df.to_string())

# Boxplot
fig, axes = plt.subplots(1, 6, figsize=(20, 5), facecolor=DARK_BG)
for i, col in enumerate(CONTINUOUS):
    ax = axes[i]
    ax.boxplot(df[col].dropna(), patch_artist=True, widths=0.5,
               boxprops=dict(facecolor=FIELD_GRN, alpha=0.6, edgecolor=FIELD_GRN),
               medianprops=dict(color="white", lw=2),
               flierprops=dict(marker="o", markerfacecolor=WARN_RED, alpha=0.6, markersize=5),
               whiskerprops=dict(color="#2E3B27"), capprops=dict(color="#2E3B27"))
    ax.set_title(col.split("(")[0].strip().upper(), fontsize=10.5)
    ax.set_facecolor(CARD_BG); ax.set_xticks([])
    n_out = outlier_mask[col].sum()
    if n_out > 0:
        ax.text(1.2, df[col].max(), f"{n_out} outliers", color=WARN_RED, fontweight="bold", fontsize=9, va="top", ha="right")

plt.tight_layout()
plt.show()
```

To preserve the sequential time-series order, we DO NOT drop rows containing outliers.

```python
# tag: yeojohnson
scaler = RobustScaler()
scaled_vals = scaler.fit_transform(df[SKEWED_FEATS := ["GDP B$", "Inflation(%)", "Temperature(°C)"]])
df_scaled = pd.DataFrame(scaled_vals, columns=SKEWED_FEATS, index=df.index)

fig, axes = plt.subplots(len(SKEWED_FEATS), 2, figsize=(14, 10), facecolor=DARK_BG)
for i, col in enumerate(SKEWED_FEATS):
    for ax_idx, (data, color, label) in enumerate([
        (df[col], GOLD_DIM, "Original"),
        (df_scaled[col], FIELD_GRN, "Robust Scaled")
    ]):
        ax = axes[i, ax_idx]
        ax.hist(data.dropna(), bins=18, color=color, alpha=0.7, edgecolor=DARK_BG, density=True)
        ax.set_title(f"{col}  —  {label}", fontsize=10, color="white")
        ax.set_facecolor(CARD_BG)
plt.tight_layout()
plt.show()
```
